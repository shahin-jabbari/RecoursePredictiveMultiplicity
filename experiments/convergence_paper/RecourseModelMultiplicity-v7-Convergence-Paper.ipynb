# V7: per-individual convergence study — training protocol v2

Six datasets, nested diverse sets, shared majority-rejected individuals, and 1,000 ORPM rounds.

**Training is checked first for all 30 dataset–seed blocks.** Logistic fits require a regularized objective-error bound; neural fits require a full-data gradient or continued-optimization stability check (neither is a global guarantee). Only after every model bank and cohort passes does the worker run convergence trajectories.

Five seeds (401–405) and five held-out individuals per seed: 4,050 trajectories. Compatible checked model fits are reused. Results are saved under `results/V7/convergence_paper/paper_training_v2_5x5/`; earlier study results are preserved. See README.txt for the full protocol.
